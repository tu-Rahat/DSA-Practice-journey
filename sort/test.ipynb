```python
# merge Sort
def divide(arr):
    if len(arr) <= 1:
        return arr
    mid = (len(arr) // 2)
    left_arr = arr[:mid]
    right_arr = arr[mid:]
    lef = divide(left_arr)
    rig = divide(right_arr) 
    return merge(lef,rig)




def merge(l,r):
    result = [] 
    i,j= 0,0
    
    # left smaller than right or equal
    while (i<len(l) or j<len(r)) and len(l)<=len(r):
        if j<len(r) and i<len(l):
            if l[i] <= r[j]:
                result.append(l[i]) 
                i = i + 1
            else:
                result.append(r[j]) 
                j = j + 1
        elif i<len(l) and j>=len(r):
            result.extend(l[i:]) 
            break
        else:
            result.extend(r[j:]) 
            break
    return result

# a = merge([1,7,9],[5,6])
# print(a)
divide([20,21,40,1,2,3,19])
```
