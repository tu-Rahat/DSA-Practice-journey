```python
# https://leetcode.com/problems/plus-one/description/?envType=problem-list-v2&envId=array

# 66 Plus One

test = [9]
l = len(test)

for i in range(l-1,-1,-1):
    if test[i] == 9:
        test[i] = 0
    else:
        test[i] = test[i] + 1
        break
else:
    test = [1] + test
print(test)
    
```

```python
# Merge Sorted Array

# https://leetcode.com/problems/merge-sorted-array/description/?envType=problem-list-v2&envId=array


"""
time = O(m+n)
space = O(1)
Iterate till nums2 len size because we just have to put those value inn nums1 vacant position
Then compare nums1 biggest value and nums2 biggest value

Ex: [1 7 8 0 0 0] = nums1 ,  [1,2,3] =nums2 (m=3,n=3, vac = 3+3-1 : total len of nums1)

As 2,3 is smaller then nums1 greater values, nums2 value have to be moved to middle;
You may think we have shift right? Possilbe but not efficient Time becomes = O(N**2)
So just copy greater values of nums1 in the vacant pos then overwrite copied valued by nums2

here 7,8 will move to last 2 idx of nums1 => [1,7,8,0,7,8] (j is still 2 but i is pointing at index 0
                                    and vac at here ^

as 2,3 are greather then 1 so  it becomes like this [1,1,2,3,7,8] 
                                                       ^vac^

and j becomes -1 loop ends :)
"""


nums1 = [1,7,8,0,0,0]
nums2 = [1,2,3]   
m = 3
n = 3
i = m - 1
j = n - 1 
vac = m + n - 1
"Print function is used for bebugging purpose"
while j>-1:
    if (nums1[i] >= nums2[j]) and i>-1:
        nums1[vac] = nums1[i] 
        i = i - 1
        vac = vac - 1
        print(f"1st if {nums1} and j={j},i={i},vac={vac}")
    elif i<0:
        nums1[vac] = nums2[j]
        
        j = j - 1
        vac = vac - 1 
        print(f"elif {nums1} and j={j},i={i},vac={vac}")
        
    else:
        nums1[vac] = nums2[j]
        
        j = j - 1
        vac = vac - 1
        print(f"else {nums1} and j={j},i={i},vac={vac}")
print(nums1)
    
```
