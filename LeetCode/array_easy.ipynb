```python
# https://leetcode.com/problems/plus-one/description/?envType=problem-list-v2&envId=array

# 66 Plus One

test = [9]
l = len(test)

for i in range(l-1,-1,-1):
    if test[i] == 9:
        test[i] = 0
    else:
        test[i] = test[i] + 1
        break
else:
    test = [1] + test
print(test)
    
```
